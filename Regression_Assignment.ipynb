**1. What is Simple Linear Regression?**

Ans:
Simple Linear Regression is a foundational statistical method used to model the relationship between two variables: one independent variable (X) and one dependent variable (Y). The goal is to find a straight-line equation that best predicts Y based on X.

**2.  What are the key assumptions of Simple Linear Regression?**

Ans:
**Assumptions:**
Linearity: The relationship between the independent variable (X) and dependent variable (Y) is linear.
Independence: Observations are independent of each other.
Homoscedasticity: The variance of residuals is constant across all levels of X.
Normality: Residuals are normally distributed.
No multicollinearity: Not applicable in SLR (since there's only one predictor), but relevant in multiple linear regression.

**3. What does the coefficient m represent in the equation Y=mX+c ?**

Ans:
In the equation Y = mX + c, the coefficient m represents the slope of the line. It indicates the rate of change of Y with respect to X

**4. What does the intercept c represent in the equation Y=mX+c?**

Ans:
In the equation Y = mX + c, the intercept c represents the value of Y when X is 0. It's the point where the line crosses the Y-axis.

**5. How do we calculate the slope m in Simple Linear Regression ?**

Ans:

The slope (m) in Simple Linear Regression is calculated using the following formula:

m = Σ[(xi - x̄)(yi - ȳ)] / Σ(xi - x̄)²

where:
xi are individual data points of the independent variable

x̄ is the mean of the independent variable

yi are individual data points of the dependent variable

ȳ is the mean of the dependent variable

**7. How is the coefficient of determination (R²) interpreted in Simple Linear Regression?**

Ans:
The coefficient of determination (R²) is a key metric in Simple Linear Regression that measures how well the regression model explains the variability in the dependent variable (Y) based on the independent variable (X).

Interpretation:

R² = 1: Perfect fit (all variance explained)
R² = 0: No relationship (no variance explained)
0 < R² < 1: Proportion of variance explained (e.g., R² = 0.7 means 70% of variance in Y is explained by X)

**9. What is the main difference between Simple and Multiple Linear Regression?**

Ans:
The main difference is the number of independent variables:

Simple Linear Regression: One independent variable (X) is used to predict the dependent variable (Y).

Multiple Linear Regression: Two or more independent variables (X1, X2, X3, etc.) are used to predict the dependent variable (Y).

**10. What are the key assumptions of Multiple Linear Regression?**

Ans:
The key assumptions of Multiple Linear Regression are:

Linearity: Relationship between independent variables and dependent variable is linear.

Independence: Observations are independent of each other.

Homoscedasticity: Constant variance of residuals across all levels of independent variables.

Normality: Residuals are normally distributed.

No multicollinearity: Independent variables are not highly correlated with each other.

**11. What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model ?**

Ans:
Heteroscedasticity refers to the condition where the variance of residuals varies across different levels of the independent variables.

This can affect the results of a Multiple Linear Regression model in several ways:

Inefficient estimates: Standard errors of coefficients may be underestimated or overestimated.
Biased statistical tests: T-statistics and F-statistics may be unreliable.
Incorrect conclusions: May lead to incorrect inferences about the relationships between variables.

**12.  How can you improve a Multiple Linear Regression model with high multicollinearity?**

Ans:
To improve a Multiple Linear Regression model with high multicollinearity, consider the following:

Remove highly correlated variables: Drop one of the correlated variables.
Use dimensionality reduction: Techniques like PCA (Principal Component Analysis) can help.
Regularization techniques: Lasso or Ridge regression can reduce the impact of multicollinearity.
Collect more data: Increasing the sample size can help alleviate multicollinearity issues.

**18. What are the limitations of using R² as a sole measure of model performance?**

Ans:
The limitations of using R² as a sole measure of model performance are:

Overfitting: R² can be inflated by adding more variables, even if they're not meaningful.
Lack of predictive power: High R² doesn't guarantee good out-of-sample predictions.
Ignores other aspects: Doesn't account for bias, variance, or other important model evaluation metrics.
Comparability issues: R² values aren't always comparable across different models or datasets.

**19. How would you interpret a large standard error for a regression coefficient?**

**20.  How can heteroscedasticity be identified in residual plots, and why is it important to address it?**

Ans:
Heteroscedasticity can be identified in residual plots by:

Non-random patterns: Residuals exhibit a funnel-shaped or cone-shaped pattern.

Unequal variance: Spread of residuals varies across different levels of predicted values.

Addressing heteroscedasticity is important because it can:

Lead to inefficient estimates
Affect statistical tests and confidence intervals

Result in incorrect conclusions

**21. What does it mean if a Multiple Linear Regression model has a high R² but low adjusted R²?**

Ans:
If a Multiple Linear Regression model has a high R² but low adjusted R², it suggests:
Overfitting: The model is too complex and fits noise in the data.
Penalty for added variables: Adjusted R² penalizes for additional predictors, revealing potential redundancy.
This discrepancy indicates that the model's high R² might not generalize well to new data, and simplification or variable selection might be necessary.

**23. What is polynomial regression?**

Ans;
Polynomial regression is a type of regression analysis where the relationship between the independent variable (X) and dependent variable (Y) is modeled using a polynomial equation.

**24. How does polynomial regression differ from linear regression?**

Ans:
Polynomial regression differs from linear regression in that it:
Allows non-linear relationships: Models curvilinear relationships between variables.
Uses higher-degree terms: Includes terms like X², X³, etc., to capture non-linear patterns.
In contrast, linear regression assumes a straight-line relationship between variables. Polynomial regression provides more flexibility to fit complex data patterns.

**25. When is polynomial regression used?**

Ans:
Polynomial regression is used when:
Non-linear relationships exist: Data exhibits curvilinear patterns.
Linear regression doesn't fit well: Relationships aren't straight-line.
Complex patterns need modeling: Captures interactions and nuances.
Common applications include:
Predicting continuous outcomes
Modeling growth curves
Analyzing complex relationships
It's particularly useful when the relationship between variables is non-linear and requires a more flexible model.

**26. What is the general equation for polynomial regression?**

Ans:
The general equation for polynomial regression is:
Y = β0 + β1X + β2X² + β3X³ + … + βnX^n + ε
Where:
Y is the dependent variable
X is the independent variable
β0, β1, β2, …, βn are coefficients
n is the degree of the polynomial
ε is the error term
This equation allows for modeling non-linear relationships between X and Y.

**27. Can polynomial regression be applied to multiple variables?**

Ans:
Yes, polynomial regression can be applied to multiple variables. This is known as multivariate polynomial regression.
The model can include:
Interaction terms: Products of different variables (e.g., X1*X2)
Polynomial terms: Higher-degree terms for each variable (e.g., X1², X2³)
This allows for modeling complex relationships between multiple independent variables and the dependent variable.

**28. What are the limitations of polynomial regression?**

Ans:
The limitations of polynomial regression are:
Overfitting: High-degree polynomials can fit noise, not just the underlying pattern.
Oscillations: Polynomials can exhibit extreme oscillations, especially at boundaries.
Lack of interpretability: Complex models can be difficult to understand.
Computational issues: High-degree polynomials can lead to numerical instability.

**31.  How is polynomial regression implemented in Python?**

Ans:

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Sample Data (Non-linear relationship)
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([2, 3, 6, 8, 12])  # Y follows a non-linear pattern

# Convert X to Polynomial Features (Degree = 2)
poly = PolynomialFeatures(degree=2)  # Can change degree for complexity
X_poly = poly.fit_transform(X)

# Train Polynomial Regression Model
model = LinearRegression()
model.fit(X_poly, Y)

# Predictions
Y_pred = model.predict(X_poly)

# Plot Results
plt.scatter(X, Y, color='blue', label="Actual Data")
plt.plot(X, Y_pred, color='red', linestyle='dashed', label="Polynomial Regression")
plt.legend()
plt.show()
```
